## Calculating sample size for review

```python
N = len(custody_data['pfa'].unique())
z = 1.96  # 95% confidence
p = 0.5   # maximum variability
e = 0.2   # 20% margin of error

n = (N * z**2 * p * (1-p)) / ((N-1)*e**2 + z**2*p*(1-p))
print(f"Recommended sample size: {math.ceil(n)}")
```

```python
random_pfas = random.sample(list(custody_data['pfa']), math.ceil(n))
print(random_pfas)
```

```python
filt = custody_data['pfa'].isin(random_pfas)
cols = list(custody_data.columns[:2]) + list(custody_data.columns[-3:])
custody_data.loc[filt, cols]
```
